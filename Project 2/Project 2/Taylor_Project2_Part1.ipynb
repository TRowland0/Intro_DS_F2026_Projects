## 7 : Filtering the Data

```python
df.drop(columns=['title'], inplace=True)
df.head()
```

```python
print('Movie ID type before and after.')
print(df['movie_id'].dtype)
df['movie_id']=df['movie_id'].astype(str)
print(df['movie_id'].dtype)
print('Blockbuster type before and after.')
print(df['blockbuster'].dtype)
df['blockbuster']=df['blockbuster'].astype(str)
print(df['blockbuster'].dtype)
```

Connecting to my earlier analysis, I converted movie_id and blockbuster to string as it makes more sense considering each value is a categorical variable. In movie ID, each ID is unique and is attached to a specific movie which makes it not measurable. For blockbuster there are only two options technically possible in the blockbuster as it's either a blockbuster or it's not.

```python
blockbuster= df[df['blockbuster']=='1']
blockbuster
```

This filter displays all movies in the dataset that have a blockbuster status.

```python
blockbuster_past_2010= df[(df['blockbuster']=='1') & (df['year']>=2010)]
blockbuster_past_2010
```

This filter displays movies in the set with a blockbuster status and they were released either in 2010 or later. <br> For the first filter: It pulls out all of the movies that have their blockbuster variable as equal to '1'. This means that all pulled movies are a blockbuster. We can then do some greater anaylsis within this to try and find correlations that these blockbusters have with each other <br> 
For the second filter: I selected movies that had a blockbuster variable equal to 1 and had a year variable >= 2010. In analysis, since it selects movies made in 2010 or newer, we can make connections and see if there is a difference in the decade. We can compare this amount of blockbusters to total blockbusters as well.
